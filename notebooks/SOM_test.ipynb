# Load data

```python
# Load prompts
pd.set_option('display.precision', 15)  # Establecer la precisión a 15 decimales

PATH = get_path(1)
DATABASE_FOLDER = PATH + 'data/processed/transcriptions.db'
```

```python
df = read_data_from_database(DATABASE_FOLDER, 'proposed_lexical_fields')

column_list = [
    'source','orthographic_word', 'lemmatized_text', 'name', 'is_in_proposed_lexical_fields', 
    'lexical_field_proposed', 'description', 'matched_word']

df[column_list].head(3)
```

```python
from pathlib import Path
from collections import defaultdict

# --- Configuración inicial (df y PATH ya definidos) ---
BASE_PATH = Path(PATH) / "data" / "ds003020"

# --- Indexación única de todos los .nii.gz (un solo paseo por disco) ---
nii_files = list(BASE_PATH.glob("sub-UTS*/**/*.nii.gz"))
print(f"Se han indexado {len(nii_files)} archivos .nii.gz")

# --- Emparejamiento por cada lexical_field_proposed con SUBSTRING match ---
matches: dict[str, list[Path]] = defaultdict(list)

for lf_value, grupo in df.groupby("lexical_field_proposed"):
    for src in grupo["source"].unique():
        # buscamos en memoria todos los archivos cuyo stem contenga src
        for nii in nii_files:
            if src in nii.stem:
                matches[lf_value].append(nii)

# --- Mostrar resultados ---
if not matches:
    print("No se encontró ningún match. Revisa los valores de 'source' y los nombres de archivos.")
else:
    for lf_value, rutas in matches.items():
        print(f"\nCampo léxico: {lf_value}")
        for ruta in rutas:
            print(f"  * {ruta}")
```
